5. Compensation

5.1 Salary Calculation

class SalaryCalculation:
    def __init__(self, monthly_salary, total_days_in_month):
        self.monthly_salary = monthly_salary
        self.total_days = total_days_in_month
    
    def calculate_lop_deduction(self, lop_days):
        """
        Calculate Loss of Pay deduction
        """
        per_day_salary = self.monthly_salary / self.total_days
        return per_day_salary * lop_days
    
    def calculate_net_salary(self, lop_days, attendance_incentive):
        """
        Calculate final salary after deductions and incentives
        """
        lop_deduction = self.calculate_lop_deduction(lop_days)
        return self.monthly_salary - lop_deduction + attendance_incentive

1. Problem Framing

    Type: Multi-class classification (3 classes: Present, Absent (Leave Approved), No Show)
    Granularity: Per employee, per day, per shift (Day / Night)
    Forecast Horizon: 7 days ahead (rolling)
    Output: Predicted class + probability for each class

2. Data Collection & Integration

Gather and unify these data sources into a single feature table:

    Attendance Records (15 months): Date, Employee ID, Shift, Attendance Status
    Leave Data (past & future): Leave applied date, leave type, approval status, leave start/end dates
    Weather Data: Temperature, precipitation, severe weather alerts (by location & date)
    Commute Data: Distance to workplace, public transport disruptions, traffic indices
    Pay Cycle: Days since last payday, days to next payday
    Local Festivals & Events: Binary flags or categorical for holidays, local events, religious festivals
    Incentive Programs: Whether the employee is currently eligible for or enrolled in attendance incentive schemes
    Employee Demographics (if available): Tenure, department, role, age band, contract type

3. Feature Engineering (Critical for Accuracy)

This is where most of the predictive power comes from:

Temporal Features:

    Day of week, week of month, month of year
    Is weekend adjacent (Monday/Friday flag)
    Is day before/after a holiday or long weekend ("sandwich day")
    Season / quarter

Employee Behavioral Features (rolling windows):

    Absence rate in last 7 / 14 / 30 / 90 days
    No-show rate in last 7 / 14 / 30 / 90 days
    Consecutive days present streak
    Average leaves per month
    Leave rejection rate (may correlate with no-shows)
    Shift-specific absence rate (Day vs Night)

Leave-Based Features:

    Has approved future leave for the target date? (strong signal)
    Has pending leave application for the target date?
    Was a leave recently rejected for the target date? (no-show risk ↑)
    Days since last leave taken

External Features:

    Weather severity score on target date
    Is it a pay day or day after pay day?
    Local festival / event flag
    Commute difficulty index

Interaction Features:

    Night shift × day-of-week (e.g., Friday night shifts may have higher absence)
    Weather severity × commute distance
    Pay cycle × day of week

4. Data Preprocessing

    Handle class imbalance: "No Show" is likely rare. Use techniques like:
        SMOTE (Synthetic Minority Oversampling)
        Class weights in the model
        Stratified sampling
    Encode categoricals: One-hot or target encoding for shift, department, day of week, etc.
    Normalize/scale continuous features (especially for distance-based models)
    Handle missing data: Impute or flag missingness as a feature itself
    Time-aware train/test split: Never leak future data into training — use a cutoff date

5. Model Selection & Training

Start simple, then increase complexity:
Table
Stage

Model

Why
Baseline

Logistic Regression

Interpretable, fast, good benchmark
Intermediate

Random Forest / LightGBM / XGBoost

Handles non-linear relationships, feature interactions, class imbalance well
Advanced

Stacked Ensemble (XGBoost + LightGBM + CatBoost)

Combines strengths of multiple learners
Optional

LSTM / Temporal Fusion Transformer

If sequential attendance patterns are very strong

Recommended primary model: LightGBM or XGBoost — they tend to perform best on tabular data with mixed feature types and handle class imbalance natively via scale_pos_weight or sample weights.
6. Validation Strategy

    Time-series cross-validation: Use expanding or sliding window CV (e.g., train on months 1–10, validate on month 11; train on 1–11, validate on 12, etc.)
    Never use random K-fold — it leaks temporal information
    Stratify by class within each fold to preserve class distribution

7. Evaluation Metrics

Since classes are imbalanced, go beyond accuracy:

    Macro F1-Score (primary metric — balances precision/recall across all 3 classes)
    Per-class Precision & Recall (especially for "No Show")
    Confusion Matrix for error analysis
    Log Loss if you need well-calibrated probabilities
    Cohen's Kappa for agreement beyond chance

8. Hyperparameter Tuning

    Use Optuna or Bayesian Optimization (more efficient than grid search)
    Key hyperparameters for XGBoost/LightGBM: learning_rate, max_depth, n_estimators, min_child_weight, subsample, colsample_bytree, scale_pos_weight

9. Model Interpretability

    Use SHAP values to explain predictions (which features drive each class)
    Generate feature importance plots
    This builds trust with stakeholders and helps validate the model isn't learning spurious patterns

10. Deployment & Forecasting Pipeline

    Weekly batch prediction: Every week, generate predictions for the next 7 days × each employee × each shift
    Retrain cadence: Monthly or bi-monthly with fresh data
    Monitor for drift: Track prediction accuracy weekly; alert if performance degrades
    Feedback loop: Capture actual outcomes to continuously improve the model

Pro Tips for Maximum Accuracy

    The "approved future leave" feature will be your strongest predictor — if someone has approved leave, they're almost certainly absent. Build the model to leverage this heavily.
    Focus feature engineering effort on the "No Show" class — it's the hardest to predict and the most operationally costly.
    Separate models per shift (Day vs Night) may outperform a single model if attendance patterns differ significantly between shifts.
    Ensemble a rules-based layer (e.g., "if leave approved → Absent") with the ML model for the uncertain cases.

```python
import pandas as pd
import numpy as np

# 1. Reload the CSV
csv_path = '/content/attendance_data/enriched_attendance.csv'
df = pd.read_csv(csv_path)

# 2. Melt to long format
date_cols = [col for col in df.columns if any(char in col for char in ['/', '.', '202']) and len(col) >= 6]
static_cols = ['Employee Code', 'Name', 'Employee ID', 'Department', 'DOJ', 'Location']
actual_static = [c for c in static_cols if c in df.columns]

df_long = pd.melt(df, id_vars=actual_static, value_vars=date_cols, var_name='Date', value_name='Attendance_Code')
df_long['Date'] = pd.to_datetime(df_long['Date'], errors='coerce')
df_long = df_long.dropna(subset=['Date'])

# 3. Map Target Classes
def map_target_class(code):
    if pd.isna(code): return 'Unknown'
    code = str(code).replace(' ', '').upper()
    present_codes = ['P', 'P-OT', 'P-WO', 'P-H', 'NP', 'NP-OT', 'P-C(MH)', 'PH', 'HPH', 'WNP']
    absent_codes = ['AL', 'CL', 'SL', 'CTO', 'LOP', 'HOLIDAY', 'SC', 'HCL', 'HLOP', 'HSL', 'H-NP', 'HOWO', 'W-NP', 'WOF', 'MLC']
    no_show_codes = ['NCNS', 'UL', 'AAP', 'RC-AL', 'RC-CL', 'RC-SL', '---']
    if code in present_codes: return 'Present'
    if code in absent_codes: return 'Absent'
    if code in no_show_codes: return 'No Show'
    return 'Unknown'

df_long['Target'] = df_long['Attendance_Code'].apply(map_target_class)

# 4. Feature Engineering
df_long = df_long.sort_values(['Employee Code', 'Date'])
df_long['DayOfWeek'] = df_long['Date'].dt.dayofweek
df_long['Shift_Type'] = df_long['Attendance_Code'].apply(lambda x: 'Night' if str(x).replace(' ','').upper() in ['NP', 'NP-OT', 'WNP', 'H-NP'] else 'Morning')
df_long['is_absent'] = (df_long['Target'] == 'Absent').astype(int)
df_long['is_no_show'] = (df_long['Target'] == 'No Show').astype(int)

grouped = df_long.groupby('Employee Code')
for days in [7, 30]:
    df_long[f'absent_rate_{days}d'] = grouped['is_absent'].transform(lambda x: x.shift().rolling(window=days, min_periods=1).mean())
    df_long[f'no_show_rate_{days}d'] = grouped['is_no_show'].transform(lambda x: x.shift().rolling(window=days, min_periods=1).mean())

print(f'Pipeline restored. df_long shape: {df_long.shape}')
```

```python
def map_target_class(code):
    if pd.isna(code): return 'Unknown'
    # Normalize code by removing spaces and converting to uppercase
    code = str(code).replace(' ', '').upper()

    # Mapping based on policy manual Section 2.3 and Section 3
    present_codes = ['P', 'P-OT', 'P-WO', 'P-H', 'NP', 'NP-OT', 'P-C(MH)', 'PH', 'HPH', 'WNP']
    absent_codes = ['AL', 'CL', 'SL', 'CTO', 'LOP', 'HOLIDAY', 'SC', 'HCL', 'HLOP', 'HSL', 'H-NP', 'HOWO', 'W-NP', 'WOF', 'MLC']
    no_show_codes = ['NCNS', 'UL', 'AAP', 'RC-AL', 'RC-CL', 'RC-SL', '---']

    if code in present_codes: return 'Present'
    if code in absent_codes: return 'Absent'
    if code in no_show_codes: return 'No Show'
    return 'Unknown'

# Apply mapping to create the target variable
df_long['Target'] = df_long['Attendance_Code'].apply(map_target_class)

# Basic Temporal Features
df_long['DayOfWeek'] = df_long['Date'].dt.dayofweek
df_long['IsWeekend'] = df_long['DayOfWeek'].apply(lambda x: 1 if x >= 5 else 0)
df_long['Month'] = df_long['Date'].dt.month

print('Target Class Distribution after mapping:')
print(df_long['Target'].value_counts(normalize=True) * 100)

display(df_long[['Employee Code', 'Date', 'Attendance_Code', 'Target']].head())
```

```python
def predict_next_7_days(employee_ids, last_date, model, df_long_full, label_encoder):
    """
    Generates attendance predictions for a list of employees for the 7 days following last_date.
    """
    import pandas as pd

    forecast_results = []
    future_dates = pd.date_range(start=last_date + pd.Timedelta(days=1), periods=7)

    # Use the last known rolling features for each employee as a starting point
    # We drop duplicates to ensure one row per employee
    latest_features = df_long_full[df_long_full['Date'] == last_date].drop_duplicates('Employee Code').set_index('Employee Code')

    for eid in employee_ids:
        if eid not in latest_features.index:
            continue

        emp_data = latest_features.loc[eid]

        for f_date in future_dates:
            feat_vector = pd.DataFrame([{
                'DayOfWeek': f_date.dayofweek,
                'Shift_Type_Enc': emp_data['Shift_Type_Enc'],
                'absent_rate_7d': emp_data['absent_rate_7d'],
                'no_show_rate_7d': emp_data['no_show_rate_7d'],
                'absent_rate_30d': emp_data['absent_rate_30d'],
                'no_show_rate_30d': emp_data['no_show_rate_30d']
            }])

            probs = model.predict_proba(feat_vector)[0]
            pred_class = model.predict(feat_vector)[0]

            forecast_results.append({
                'Employee Code': eid,
                'Date': f_date.date(),
                'Predicted_Status': label_encoder.inverse_transform([pred_class])[0],
                'Prob_Present': probs[2],
                'Prob_Absent': probs[0],
                'Prob_No_Show': probs[1]
            })

    return pd.DataFrame(forecast_results)

# Using the variables created in the consolidated pipeline cell
if 'df_long' in globals() and 'clf' in globals():
    sample_employees = df_long['Employee Code'].unique()[:5]
    last_known_date = df_long['Date'].max()

    forecast_df = predict_next_7_days(sample_employees, last_known_date, clf, df_long, le)
    print(f'\nAttendance Forecast for {last_known_date + pd.Timedelta(days=1)} to {last_known_date + pd.Timedelta(days=7)}:')
    display(forecast_df.head(10))
else:
    print('Error: Pipeline variables (df_long, clf) not found in memory. Please re-run the training cell.')
```

```python
def generate_production_forecast(model, df_long_full, label_encoder, days=7):
    """
    Final Phase 10 Pipeline: Generates a 7-day forecast for the entire workforce.
    """
    last_date = df_long_full['Date'].max()
    future_dates = pd.date_range(start=last_date + pd.Timedelta(days=1), periods=days)

    # Get latest behavioral stats for all employees
    latest_stats = df_long_full[df_long_full['Date'] == last_date].drop_duplicates('Employee Code')

    all_preds = []

    for f_date in future_dates:
        batch_feats = latest_stats.copy()
        batch_feats['Date'] = f_date
        batch_feats['DayOfWeek'] = f_date.dayofweek
        batch_feats['is_monday'] = 1 if f_date.dayofweek == 0 else 0
        batch_feats['is_friday'] = 1 if f_date.dayofweek == 4 else 0

        # Define feature columns used in training
        feat_cols = ['DayOfWeek', 'absent_rate_7d', 'no_show_rate_7d', 'absent_rate_30d', 'no_show_rate_30d', 'is_monday', 'is_friday']

        X_batch = batch_feats[feat_cols].fillna(0)
        preds = model.predict(X_batch)
        probs = model.predict_proba(X_batch)

        batch_feats['Predicted_Status'] = label_encoder.inverse_transform(preds)
        # Probability of 'No Show' is usually index 1 if classes are ['Absent', 'No Show', 'Present']
        batch_feats['No_Show_Risk_Score'] = probs[:, 1]

        all_preds.append(batch_feats[['Employee Code', 'Date', 'Predicted_Status', 'No_Show_Risk_Score']])

    final_forecast = pd.concat(all_preds)
    return final_forecast

# Run the automated pipeline
if 'lgbm' in locals():
    full_forecast = generate_production_forecast(lgbm, df_long, le)
    print(f'Pipeline complete. Generated {len(full_forecast)} predictions for the next 7 days.')
    display(full_forecast.sort_values('No_Show_Risk_Score', ascending=False).head(10))
else:
    print('Please ensure the LightGBM model is trained before running the pipeline.')
```

### Phase 6: Time-Series Cross-Validation
To properly validate the model, we must respect the temporal order of events. We will use a sliding window approach where the model is trained on a set period (e.g., 12 months) and validated on the subsequent week, repeating this process over multiple folds.

```python
from sklearn.model_selection import TimeSeriesSplit

def run_timeseries_cv(X, y, model):
    tscv = TimeSeriesSplit(n_splits=5)
    scores = []

    print('Starting Time-Series Cross-Validation...')
    for train_index, test_index in tscv.split(X):
        X_train_cv, X_test_cv = X.iloc[train_index], X.iloc[test_index]
        y_train_cv, y_test_cv = y[train_index], y[test_index]

        model.fit(X_train_cv, y_train_cv)
        score = model.score(X_test_cv, y_test_cv)
        scores.append(score)
        print(f'Fold Score: {score:.4f}')

    print(f'\nAverage CV Accuracy: {np.mean(scores):.4f}')

# Note: This requires the balanced X and y from the training cell to be fully populated.
```

### Phase 6 Implementation: Executing Chronological Validation
We use `TimeSeriesSplit` to ensure that for every fold, the training set occurs before the validation set in time. This provides a realistic estimate of how the model will perform on future data.

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import os
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import f1_score
from sklearn.preprocessing import LabelEncoder

# 1. Load data if missing
if 'df_long' not in locals():
    path = '/content/attendance_data_full_long.parquet'
    if os.path.exists(path):
        print('Loading data from parquet...')
        df_long = pd.read_parquet(path)
    else:
        print('Error: Parquet file not found. Please run the export cell (e26b1e2e) first.')

# 2. Re-prepare the balanced X and y from the processed data
if 'df_long' in locals():
    print("Reconstructing training data for validation...")
    features_v3 = ['DayOfWeek', 'absent_rate_7d', 'no_show_rate_7d', 'absent_rate_30d', 'no_show_rate_30d', 'is_sandwich_day', 'is_tier1_eligible']

    # Filtering and sampling for validation performance
    df_model_cv = df_long[df_long['Target'] != 'Unknown'].copy()
    present_sample = df_model_cv[df_model_cv['Target'] == 'Present'].sample(frac=0.05, random_state=42)
    others = df_model_cv[df_model_cv['Target'] != 'Present']
    df_train_cv = pd.concat([present_sample, others]).sort_values('Date')

    X = df_train_cv[features_v3].fillna(0)
    le = LabelEncoder()
    y = le.fit_transform(df_train_cv['Target'])

    # 3. Run Time-Series Cross-Validation
    tscv = TimeSeriesSplit(n_splits=3)
    f1_scores = []

    print("Starting Time-Series Cross-Validation on Balanced Dataset...\n")

    for i, (train_index, val_index) in enumerate(tscv.split(X)):
        X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
        y_train_fold, y_val_fold = y[train_index], y[val_index]

        cv_model = lgb.LGBMClassifier(
            n_estimators=50,
            learning_rate=0.1,
            class_weight='balanced',
            random_state=42
        )

        cv_model.fit(X_train_fold, y_train_fold)
        y_pred_fold = cv_model.predict(X_val_fold)

        score = f1_score(y_val_fold, y_pred_fold, average='macro')
        f1_scores.append(score)
        print(f"Fold {i+1} - Macro F1 Score: {score:.4f}")

    print(f"\nAverage Time-Series Macro F1: {np.mean(f1_scores):.4f}")
```

```python
# 2. Generate Roster Risk Assessment (Top 20 High-Risk Employees)
def generate_risk_report(model, df_long_full, label_encoder, days=7):
    last_date = df_long_full['Date'].max()
    future_dates = pd.date_range(start=last_date + pd.Timedelta(days=1), periods=days)

    # Get the most recent behavioral profile for each employee
    latest_profiles = df_long_full[df_long_full['Date'] == last_date].drop_duplicates('Employee Code')

    risk_list = []
    for _, emp in latest_profiles.iterrows():
        # Prepare features for prediction (assuming Monday/Friday flags based on future dates)
        for f_date in future_dates:
            feats = pd.DataFrame([{
                'DayOfWeek': f_date.dayofweek,
                'absent_rate_7d': emp['absent_rate_7d'],
                'no_show_rate_7d': emp['no_show_rate_7d'],
                'absent_rate_30d': emp['absent_rate_30d'],
                'no_show_rate_30d': emp['no_show_rate_30d'],
                'is_monday': 1 if f_date.dayofweek == 0 else 0,
                'is_friday': 1 if f_date.dayofweek == 4 else 0
            }])

            probs = model.predict_proba(feats)[0]
            no_show_prob = probs[np.where(label_encoder.classes_ == 'No Show')[0][0]]

            risk_list.append({
                'Employee Code': emp['Employee Code'],
                'Date': f_date.date(),
                'No_Show_Probability': no_show_prob
            })

    return pd.DataFrame(risk_list).sort_values('No_Show_Probability', ascending=False)

if 'lgbm' in locals():
    risk_report = generate_risk_report(lgbm, df_long, le)
    print('\n--- Top 20 High-Risk Attendance Alerts (Next 7 Days) ---')
    display(risk_report.head(20))
else:
    print('Model not ready for risk assessment.')
```

```python
import zipfile
import re

def brute_force_docx_text(docx_path):
    print(f'--- Brute Force Extraction: {docx_path} ---')
    try:
        with zipfile.ZipFile(docx_path, 'r') as z:
            # List all files inside the docx
            all_files = z.namelist()
            found_any_text = False

            for file in all_files:
                if file.endswith('.xml'):
                    content = z.read(file).decode('utf-8', errors='ignore')
                    # Find text between <w:t> tags
                    text_matches = re.findall(r'<w:t[^>]*>(.*?)</w:t>', content)
                    if text_matches:
                        clean_text = ' '.join(text_matches)
                        if clean_text.strip():
                            print(f'\n[Found in {file}]:')
                            print(clean_text)
                            found_any_text = True

            if not found_any_text:
                print('No text content found in any XML sub-file.')
    except Exception as e:
        print(f'Error: {e}')

brute_force_docx_text('/content/Document(1).docx')
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import os
import zipfile
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# 1. Extraction
zip_path = '/content/attendance_with_leave_history.zip'
extract_path = '/content/attendance_data'
csv_path = '/content/attendance_data/enriched_attendance.csv'
parquet_path = '/content/attendance_data_full_long.parquet'

if os.path.exists(zip_path):
    print('--- Step 1: Extracting Source Data ---')
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)

    # 2. Backbone Reconstruction
    print('--- Step 2: Reconstructing 38M Row Backbone ---')
    df = pd.read_csv(csv_path)
    date_cols = [col for col in df.columns if any(char in col for char in ['/', '.', '202']) and len(col) >= 6]
    static_cols = ['Employee Code', 'Location']
    df_long = pd.melt(df, id_vars=static_cols, value_vars=date_cols, var_name='Date', value_name='Attendance_Code')
    df_long['Date'] = pd.to_datetime(df_long['Date'], dayfirst=True, errors='coerce')
    df_long = df_long.dropna(subset=['Date']).sort_values(['Employee Code', 'Date'])

    def map_target(code):
        c = str(code).replace(' ', '').upper()
        if c in ['P', 'NP', 'P-OT', 'NP-OT', 'WNP', 'PH']: return 'Present'
        if c in ['AL', 'CL', 'SL', 'WOF', 'HOLIDAY', 'CTO', 'LOP']: return 'Absent'
        return 'No Show'

    df_long['Target'] = df_long['Attendance_Code'].apply(map_target)
    df_long['is_absent'] = (df_long['Target'] == 'Absent').astype(int)
    df_long['is_no_show'] = (df_long['Target'] == 'No Show').astype(int)

    # Feature Engineering
    grouped = df_long.groupby('Employee Code')
    df_long['absent_rate_30d'] = grouped['is_absent'].transform(lambda x: x.shift().rolling(30, min_periods=1).mean())
    df_long['no_show_rate_30d'] = grouped['is_no_show'].transform(lambda x: x.shift().rolling(30, min_periods=1).mean())
    df_long['is_off'] = df_long['Attendance_Code'].astype(str).str.contains('WOF|Holiday', case=False, na=False).astype(int)
    df_long['is_sandwich_day'] = ((df_long['is_off'] == 0) & (grouped['is_off'].shift(1) == 1) & (grouped['is_off'].shift(-1) == 1)).astype(int)
    df_long['is_tier1_eligible'] = ((df_long['absent_rate_30d'] * 30 <= 2) & (df_long['no_show_rate_30d'] == 0)).astype(int)
    df_long['DayOfWeek'] = df_long['Date'].dt.dayofweek

    df_long.to_parquet(parquet_path, index=False)
    print(f'Backbone saved to Parquet: {len(df_long):,} rows.')

    # 3. Training
    print('--- Step 3: Training Balanced LightGBM ---')
    df_model = df_long[df_long['Target'] != 'Unknown'].copy()
    present_sample = df_model[df_model['Target'] == 'Present'].sample(frac=0.01, random_state=42)
    others = df_model[df_model['Target'] != 'Present']
    df_train = pd.concat([present_sample, others])

    features = ['DayOfWeek', 'absent_rate_30d', 'no_show_rate_30d', 'is_sandwich_day', 'is_tier1_eligible']
    le = LabelEncoder()
    y = le.fit_transform(df_train['Target'])
    X = df_train[features].fillna(0)

    lgbm = lgb.LGBMClassifier(n_estimators=100, class_weight='balanced', random_state=42)
    lgbm.fit(X, y)

    # 4. Forecast
    print('--- Step 4: Generating Risk Forecast ---')
    last_date = df_long['Date'].max()
    latest_profiles = df_long[df_long['Date'] == last_date].drop_duplicates('Employee Code').copy()
    no_show_idx = np.where(le.classes_ == 'No Show')[0][0]

    X_forecast = latest_profiles[features].fillna(0)
    latest_profiles['No_Show_Probability'] = lgbm.predict_proba(X_forecast)[:, no_show_idx]

    print('\n✅ Success! Top 10 High-Risk Employees for next shift:')
    display(latest_profiles[['Employee Code', 'Location', 'No_Show_Probability']].sort_values('No_Show_Probability', ascending=False).head(10))
else:
    print('❌ Error: attendance_with_leave_history.zip not found in /content/. Please upload it.')
```

```python
import pandas as pd
import numpy as np

def optimize_dataframe(df):
    """Reduce memory usage by downcasting types and using categories."""
    start_mem = df.memory_usage().sum() / 1024**2

    for col in df.columns:
        col_type = df[col].dtype

        if col_type == object:
            df[col] = df[col].astype('category')
        elif 'int' in str(col_type):
            df[col] = pd.to_numeric(df[col], downcast='integer')
        elif 'float' in str(col_type):
            df[col] = pd.to_numeric(df[col], downcast='float')

    end_mem = df.memory_usage().sum() / 1024**2
    print(f'Memory reduced from {start_mem:.2f} MB to {end_mem:.2f} MB ({100 * (start_mem - end_mem) / start_mem:.1f}% reduction)')
    return df

# Configuration for Efficient Loading
print("System Optimization: Ready to apply Downcasting and Category encoding to df_long.")
```

```python
import pandas as pd
import numpy as np
import lightgbm as lgb
from sklearn.preprocessing import LabelEncoder
import os
import zipfile

# Paths
zip_path = '/content/attendance_with_leave_history.zip'
extract_path = '/content/attendance_data'
csv_path = '/content/attendance_data/enriched_attendance.csv'
parquet_path = '/content/attendance_data_full_long.parquet'

# 1. Reconstruct Backbone if missing
if not os.path.exists(parquet_path):
    print('--- Reconstructing Data Backbone ---')
    if not os.path.exists(csv_path):
        if os.path.exists(zip_path):
            with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                zip_ref.extractall(extract_path)
        else:
            raise FileNotFoundError('Source ZIP file not found. Please upload attendance_with_leave_history.zip')

    df = pd.read_csv(csv_path)
    date_cols = [col for col in df.columns if any(char in col for char in ['/', '.', '202']) and len(col) >= 6]
    static_cols = ['Employee Code', 'Location']
    df_long = pd.melt(df, id_vars=static_cols, value_vars=date_cols, var_name='Date', value_name='Attendance_Code')
    df_long['Date'] = pd.to_datetime(df_long['Date'], dayfirst=True, errors='coerce')
    df_long = df_long.dropna(subset=['Date']).sort_values(['Employee Code', 'Date'])

    def map_target(code):
        c = str(code).replace(' ', '').upper()
        if c in ['P', 'NP', 'P-OT', 'NP-OT', 'WNP']: return 'Present'
        if c in ['AL', 'CL', 'SL', 'WOF', 'HOLIDAY']: return 'Absent'
        return 'No Show'

    df_long['Target'] = df_long['Attendance_Code'].apply(map_target)
    df_long['is_absent'] = (df_long['Target'] == 'Absent').astype(int)
    df_long['is_no_show'] = (df_long['Target'] == 'No Show').astype(int)

    grouped = df_long.groupby('Employee Code')
    df_long['absent_rate_30d'] = grouped['is_absent'].transform(lambda x: x.shift().rolling(30, min_periods=1).mean())
    df_long['no_show_rate_30d'] = grouped['is_no_show'].transform(lambda x: x.shift().rolling(30, min_periods=1).mean())
    df_long['is_off'] = df_long['Attendance_Code'].astype(str).str.contains('WOF|Holiday', case=False, na=False).astype(int)
    df_long['is_sandwich_day'] = ((df_long['is_off'] == 0) & (grouped['is_off'].shift(1) == 1) & (grouped['is_off'].shift(-1) == 1)).astype(int)
    df_long['is_tier1_eligible'] = ((df_long['absent_rate_30d'] * 30 <= 2) & (df_long['no_show_rate_30d'] == 0)).astype(int)
    df_long['DayOfWeek'] = df_long['Date'].dt.dayofweek

    df_long.to_parquet(parquet_path, index=False)
    print(f'Backbone saved: {len(df_long):,} rows.')
else:
    print('Loading existing backbone...')
    df_long = pd.read_parquet(parquet_path)

# 2. Training
print('--- Training Balanced LightGBM ---')
features = ['DayOfWeek', 'absent_rate_30d', 'no_show_rate_30d', 'is_sandwich_day', 'is_tier1_eligible']
df_model = df_long[df_long['Target'] != 'Unknown'].copy()
present_sample = df_model[df_model['Target'] == 'Present'].sample(frac=0.01, random_state=42)
others = df_model[df_model['Target'] != 'Present']
df_train = pd.concat([present_sample, others])

le = LabelEncoder()
y = le.fit_transform(df_train['Target'])
X = df_train[features].fillna(0)

lgbm = lgb.LGBMClassifier(n_estimators=100, class_weight='balanced', random_state=42)
lgbm.fit(X, y)

print('✅ Phase 5 Restoration SUCCESSFUL.')
```
